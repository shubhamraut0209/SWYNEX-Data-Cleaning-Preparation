# **🟨 SECTION 12 — Examine Date Formats**

```python
print("Sample Order_Date values:")

df["Order_Date"].head(30).tolist()
```

```python
import re
from collections import Counter

def identify_date_format(value):
    value = str(value)

    if re.match(r"^\d{4}-\d{2}-\d{2}$", value):
        return "YYYY-MM-DD"

    elif re.match(r"^\d{2}-\d{2}-\d{4}$", value):
        return "MM-DD-YYYY"

    elif re.match(r"^\d{2}/\d{2}/\d{4}$", value):
        return "DD/MM/YYYY"

    elif re.match(r"^\d{4}/\d{2}/\d{2}$", value):
        return "YYYY/MM/DD"

    elif re.match(r"^\d{2} [A-Za-z]{3} \d{4}$", value):
        return "DD Mon YYYY"

    else:
        return "Unknown"

date_format_counts = (
    df["Order_Date"]
    .apply(identify_date_format)
    .value_counts()
)

date_format_counts
```

# **🟨 SECTION 13 — Create Cleaning Copy**

# Data Cleaning

A copy of the original dataset is created before cleaning so that the
raw dataset remains unchanged.

```python
cleaned_df = df.copy()

print("Cleaning copy created.")
```

# **🟨 SECTION 14 — Standardize Column Names**

```python
cleaned_df.columns = (
    cleaned_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

cleaned_df.columns
```

# **🟨 SECTION 15 — Remove Exact Duplicate Records**

```python
duplicates_before = cleaned_df.duplicated().sum()

print("Duplicate records found:", duplicates_before)
```

```python
cleaned_df = cleaned_df.drop_duplicates().reset_index(drop=True)
```

```python
duplicates_after = cleaned_df.duplicated().sum()

print("Duplicate records remaining:", duplicates_after)
```

# **🟨 SECTION 16 — Clean Text Whitespace**

```python
text_columns = cleaned_df.select_dtypes(
    include="object"
).columns

for column in text_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()
```

# **🟨 SECTION 17 — Standardize Payment Methods**

```python
payment_mapping = {
    "paypal": "PayPal",
    "Pay Pal": "PayPal",
    "PayPal": "PayPal",

    "upi": "UPI",
    "U.P.I": "UPI",
    "UPI": "UPI",

    "credit card": "Credit Card",
    "CREDIT CARD": "Credit Card",
    "Credit_Card": "Credit Card",
    "Credit Card": "Credit Card",

    "debit card": "Debit Card",
    "DEBIT CARD": "Debit Card",
    "Debit Card": "Debit Card",

    "net banking": "Net Banking",
    "NetBanking": "Net Banking",
    "Net Banking": "Net Banking",

    "cod": "Cash on Delivery",
    "COD": "Cash on Delivery",
    "cash on delivery": "Cash on Delivery",
    "Cash on Delivery": "Cash on Delivery"
}

cleaned_df["payment_method"] = (
    cleaned_df["payment_method"]
    .replace(payment_mapping)
)
```

```python
cleaned_df["payment_method"].value_counts()
```

# **🟨 SECTION 18 — Standardize Countries**

```python
country_mapping = {
    "IN": "India",
    "India": "India",
    "india": "India",

    "US": "United States",
    "USA": "United States",
    "U.S.A": "United States",
    "United States": "United States",

    "UK": "United Kingdom",
    "U.K.": "United Kingdom",
    "United Kingdom": "United Kingdom",

    "CA": "Canada",
    "Canada": "Canada",
    "canada": "Canada",

    "AU": "Australia",
    "Australia": "Australia",
    "australia": "Australia",

    "DE": "Germany",
    "Germany": "Germany",
    "germany": "Germany",

    "UAE": "United Arab Emirates",
    "U.A.E": "United Arab Emirates",
    "uae": "United Arab Emirates"
}

cleaned_df["country"] = (
    cleaned_df["country"]
    .replace(country_mapping)
)
```

```python
cleaned_df["country"].value_counts()
```

# **🟨 SECTION 19 — Standardize Order Dates**

```python
def parse_order_date(value):
    value = str(value).strip()

    formats = [
        "%Y-%m-%d",
        "%m-%d-%Y",
        "%d/%m/%Y",
        "%Y/%m/%d",
        "%d %b %Y"
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt)
        except ValueError:
            continue

    return pd.NaT
```

```python
cleaned_df["order_date"] = (
    cleaned_df["order_date"]
    .apply(parse_order_date)
)
```

```python
print(cleaned_df["order_date"].dtype)
```

# **🟨 SECTION 20 — Check for Unparsed Dates**

```python
invalid_dates = cleaned_df[
    cleaned_df["order_date"].isna()
]

print("Unparsed dates:", len(invalid_dates))
```

# **🟨 SECTION 21 — Clean Numeric Data Types**

```python
numeric_columns = [
    "quantity",
    "unit_price_usd",
    "discount_percent",
    "customer_rating"
]

for column in numeric_columns:
    cleaned_df[column] = pd.to_numeric(
        cleaned_df[column],
        errors="coerce"
    )
```

```python
cleaned_df[numeric_columns].dtypes
```

# **🟨 SECTION 22 — Handle Invalid Quantities**

```python
invalid_quantity_count = (
    cleaned_df["quantity"] <= 0
).sum()

print(
    "Invalid quantity records:",
    invalid_quantity_count
)
```

```python
cleaned_df = cleaned_df[
    cleaned_df["quantity"] > 0
].reset_index(drop=True)
```

```python
print(
    "Remaining invalid quantities:",
    (cleaned_df["quantity"] <= 0).sum()
)
```

# **🟨 SECTION 23 — Handle Missing Values**

```python
cleaned_df.isnull().sum()
```

**Customer Rating**

```python
print(
    "Missing Customer Ratings:",
    cleaned_df["customer_rating"].isna().sum()
)
```

 ***Customer Rating***

Missing customer ratings were retained as null values because a missing
rating indicates that no rating was provided. Replacing these values with
0 would incorrectly imply an actual customer rating of zero.

**Shipping City**

```python
print(
    "Missing Shipping Cities:",
    cleaned_df["shipping_city"].isna().sum()
)
```

***Shipping City***

Missing shipping-city values were retained because the correct city
cannot be reliably inferred from the available fields.

**Discount**

```python
cleaned_df["discount_percent"] = (
    cleaned_df["discount_percent"].fillna(0)
)
```

```python
print(
    "Missing discounts remaining:",
    cleaned_df["discount_percent"].isna().sum()
)
```

***Discount Percentage***

Missing discount percentages were replaced with 0 because the cleaning
assumption is that a blank discount indicates that no discount was applied.
This assumption is documented rather than silently imputing a value.
